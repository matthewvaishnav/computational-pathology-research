## Quick Training Example

```python
import histocore

# Train nnMIL on PatchCamelyon (5 epochs for demo)
results = histocore.quick_train(
    dataset="pcam",
    model="nnmil", 
    epochs=5,
    batch_size=32
)

print(f"✅ Training complete!")
print(f"📈 Best accuracy: {results['best_accuracy']:.3f}")
print(f"📈 Best AUC: {results['best_auc']:.3f}")
```

## Model Comparison

```python
# Compare different MIL architectures
models = ['nnmil', 'attention', 'clam']
results = {}

for model in models:
    print(f"🚀 Training {model}...")
    result = histocore.quick_train(
        dataset="pcam",
        model=model,
        epochs=3  # Quick demo
    )
    results[model] = result
    print(f"✅ {model}: {result['best_accuracy']:.3f} accuracy")
```

## Visualization

```python
import matplotlib.pyplot as plt
import pandas as pd

# Create comparison plot
df = pd.DataFrame({
    'Model': list(results.keys()),
    'Accuracy': [r['best_accuracy'] for r in results.values()],
    'AUC': [r['best_auc'] for r in results.values()]
})

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Accuracy comparison
ax1.bar(df['Model'], df['Accuracy'])
ax1.set_title('Model Accuracy Comparison')
ax1.set_ylabel('Accuracy')
ax1.set_ylim(0, 1)

# AUC comparison
ax2.bar(df['Model'], df['AUC'])
ax2.set_title('Model AUC Comparison')
ax2.set_ylabel('AUC')
ax2.set_ylim(0, 1)

plt.tight_layout()
plt.show()

print("📊 Results Summary:")
print(df.round(3))
```

## Foundation Model Integration

```python
# Load foundation model for feature extraction
foundation_model = histocore.load_foundation_model('phikon')

# Train with foundation model features
results_foundation = histocore.quick_train(
    dataset="pcam",
    model="nnmil",
    feature_extractor=foundation_model,
    epochs=3
)

print(f"🔬 Foundation model results:")
print(f"📈 Accuracy: {results_foundation['best_accuracy']:.3f}")
print(f"📈 AUC: {results_foundation['best_auc']:.3f}")
```

## Benchmark Against Published Results

```python
# Run comprehensive benchmark
benchmark_results = histocore.benchmark(
    model_name="histocore_nnmil",
    output_dir="results/"
)

print("🏆 Benchmark Results:")
for metric, value in benchmark_results.items():
    print(f"📊 {metric}: {value}")
```
